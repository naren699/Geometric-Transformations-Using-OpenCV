# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
# Import
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image using OpenCV
img = cv2.imread('/content/baseball.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
h, w = img.shape[:2]
```

```python
# Create translation matrix
M = np.float32([
    [1, 0, 50],
    [0, 1, 80]
])
# Apply translation
translated = cv2.warpAffine(img, M, (w, h))
```

```python
# Display the image using Matplotlib
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(translated)
plt.title("Translated Image")
plt.axis("off")
plt.show()
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
# Horizontal shearing
M1 = np.float32([
    [1, 0.5, 0],
    [0, 1, 0]
])
horizontal_shear = cv2.warpAffine(
    img, M1, (w + int(0.5*h), h)
)
# Vertical shearing
M2 = np.float32([
    [1, 0, 0],
    [0.5, 1, 0]
])
vertical_shear = cv2.warpAffine(
    img, M2, (w, h + int(0.5*w))
)
```

```python
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 3, 2)
plt.imshow(horizontal_shear)
plt.title("Horizontal Shearing")
plt.axis("off")
plt.subplot(1, 3, 3)
plt.imshow(vertical_shear)
plt.title("Vertical Shearing")
plt.axis("off")
plt.show()
```

# **Image Reflection**

Write a Python program using OpenCV to perform image reflection on the **given image**.

The program should:

1. Generate the following reflected versions of the image:
   - **Horizontal Reflection** (flip left ↔ right)  
   - **Vertical Reflection** (flip top ↔ bottom)  
   - **Both-axis Reflection**  
2. Use `cv2.flip` with appropriate flip codes.
3. Display the following using a **2 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Reflected Image**  
   - **Vertically Reflected Image**  
   - **Both-axis Reflected Image**


```python
# Horizontal reflection
horizontal_flip = cv2.flip(img, 1)
# Vertical reflection
vertical_flip = cv2.flip(img, 0)
# Both-axis reflection
both_flip = cv2.flip(img, -1)
```

```python
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.subplot(2, 2, 2)
plt.imshow(horizontal_flip)
plt.title("Horizontal Reflection")
plt.axis("off")
plt.subplot(2, 2, 3)
plt.imshow(vertical_flip)
plt.title("Vertical Reflection")
plt.axis("off")
plt.subplot(2, 2, 4)
plt.imshow(both_flip)
plt.title("Both-axis Reflection")
plt.axis("off")
plt.tight_layout()
plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
# Rotation by 45 degrees
M1 = cv2.getRotationMatrix2D(
    (w//2, h//2), 45, 1
)
rotated_45 = cv2.warpAffine(img, M1, (w, h))
# Rotation by 90 degrees
M2 = cv2.getRotationMatrix2D(
    (w//2, h//2), 90, 1
)
rotated_90 = cv2.warpAffine(img, M2, (w, h))
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(rotated_45)
plt.title("45 Degree Rotation")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(rotated_90)
plt.title("90 Degree Rotation")
plt.axis("off")

plt.show()
```
